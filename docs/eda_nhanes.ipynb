# EDA - NHANES (Fase 3)
Cuaderno de evidencia: estadistica descriptiva, visualizacion, correlacion y hallazgos.

```python
%matplotlib inline
import sys
from pathlib import Path
sys.path.insert(0, str(Path().resolve().parent / 'src'))
import pandas as pd
from servicios.eda import AnalizadorDescriptivo, Visualizador, AnalizadorCorrelacion
df = pd.read_csv(Path().resolve().parent / 'outputs/cleaned/nhanes_master_cleaned.csv')
nums = ['RIDAGEYR','BMXWT','BMXHT','BMXBMI','BMXWAIST','LBXGLU','LBXTC','LBDLDL','LBXTR','DR1TKCAL','DR1TSUGR','DR1TPROT','DR1TCARB','DR1TFIBE','DR1TTFAT']
nums = [c for c in nums if c in df.columns]
desc = AnalizadorDescriptivo(nums).resumen(df)
print(desc[['mean','50%','std','skew','kurtosis']].round(2).to_string())
out = AnalizadorDescriptivo(nums).outliersIQR(df)
print(out.to_string(index=False))
```

```python
figs = Path().resolve().parent / 'outputs/figuras'
viz = Visualizador(figs)
for col in ['RIDAGEYR','BMXBMI','LBXGLU','LBXTC','DR1TKCAL']:
    if col in df.columns:
        viz.histograma(df, col)
        viz.boxplot(df, col)
for col in ['RIAGENDR','RIDRETH1','DMDEDUC2','HUQ010']:
    if col in df.columns:
        viz.barras(df, col)
if 'RIDAGEYR' in df.columns and 'BMXBMI' in df.columns:
    viz.scatter(df, 'RIDAGEYR', 'BMXBMI')
viz.heatmap(df, nums)
viz.matrizDispersion(df, nums, maxCols=6)
```

```python
corr = AnalizadorCorrelacion(nums)
pares = corr.paresDestacados(df, umbral=0.5)
print(pares.to_string(index=False))
```
